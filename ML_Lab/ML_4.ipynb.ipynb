```python
# Pick every text (object) column
text_cols = df.select_dtypes(include='object').columns

# For each one: turn line breaks into spaces, then trim spaces from both ends
for c in text_cols:
    df[c] = df[c].str.replace('\n', ' ', regex=False).str.strip()
```

```python
cols = ['Value', 'Wage', 'Release Clause']
print(df[cols].head())
print(df[cols].dtypes)
```

```python
def convert_amount(x):
    """Convert text like '€67.5M', '€220K' or '€0' into a plain number."""
    if pd.isna(x):                      # keep missing cells as NaN
        return np.nan
    x = str(x).replace('€', '').strip() # remove the euro sign
    if x.endswith('M'):
        return round(float(x[:-1]) * 1_000_000)   # x[:-1] drops the last letter
    elif x.endswith('K'):
        return round(float(x[:-1]) * 1_000)
    else:
        return float(x)                 # plain number like '0'
```

```python
for c in cols:
    df[c] = df[c].apply(convert_amount)

print(df[cols].head())
print(df[cols].dtypes)
```

```python
# Look at the 40 feet/inches rows before changing anything
mask_cm = df['Height'].str.endswith('cm')
print(df.loc[~mask_cm, 'Height'].head())
print(df['Height'].dtype)
```

```python
# 1. Rows already in cm: pull out the digits before 'cm'
cm = df['Height'].str.extract(r'(\d+)cm')[0].astype(float)

# 2. Rows in feet/inches: pull out feet and inches as two separate numbers
ft_in = df['Height'].str.extract(r"(\d+)'(\d+)")
feet = ft_in[0].astype(float)
inches = ft_in[1].astype(float)

# 3. Convert feet/inches to cm using the formula from the assignment
ft_cm = feet * 30.48 + inches * 2.54

# 4. Use the cm value where it exists, otherwise the converted one; round to 1 decimal
df['Height'] = cm.fillna(ft_cm).round(1)
```

```python
print(df['Height'].head())
print(df['Height'].dtype)
print("Missing:", df['Height'].isnull().sum())
print("Min:", df['Height'].min(), "Max:", df['Height'].max())
```

```python
# Look at the lbs rows before changing anything
is_lbs = df['Weight'].str.endswith('lbs')
print(df.loc[is_lbs, 'Weight'].head())
print(df['Weight'].dtype)
```

```python
# True for rows stored in pounds
is_lbs = df['Weight'].str.endswith('lbs')

# Pull the digits out of each format (the other format gives NaN)
kg  = df['Weight'].str.extract(r'(\d+)kg')[0].astype(float)
lbs = df['Weight'].str.extract(r'(\d+)lbs')[0].astype(float)

# If the row is in lbs, convert with lbs * 0.453592; otherwise keep the kg value
df['Weight'] = np.where(is_lbs, lbs * 0.453592, kg).round(1)
```

```python
print(df['Weight'].head())
print(df['Weight'].dtype)
print("Missing:", df['Weight'].isnull().sum())
print("Min:", df['Weight'].min(), "Max:", df['Weight'].max())
```

```python
star_cols = ['W/F', 'SM', 'IR']
print(df[star_cols].head())
print(df[star_cols].dtypes)
print(df['W/F'].unique())
```

```python
for c in star_cols:
    # Grab the first digit (r'(\d)') and drop the star symbol
    df[c] = df[c].str.extract(r'(\d)')[0]
    # Turn the extracted text digit into a real number
    df[c] = pd.to_numeric(df[c])
```

```python
print(df['Joined'].head().tolist())
print(df['Joined'].dtype)
```

```python
# %b = short month name (Jul), %d = day, %Y = 4-digit year
df['Joined'] = pd.to_datetime(df['Joined'], format='%b %d, %Y')
```

```python
# Reference date: 1 January 2021 (the assignment's choice)
ref_date = pd.Timestamp('2021-01-01')

# Difference in days, divided by 365.25 to get years (365.25 accounts for leap years)
df['Years_at_Club'] = ((ref_date - df['Joined']).dt.days / 365.25).round(1)
```

```python
print(df[['Joined', 'Years_at_Club']].head())
print(df['Joined'].dtype)
print("Missing:", df['Joined'].isnull().sum())

# Players with more than 10 years at their club
long_serving = df[df['Years_at_Club'] > 10][['Name', 'Club', 'Joined', 'Years_at_Club']]
print("Players with more than 10 years:", len(long_serving))
long_serving.sort_values('Years_at_Club', ascending=False)
```

```python
#Outliers
# Q1 = 25% mark, Q3 = 75% mark of the Wage values
q1, q3 = df['Wage'].quantile([0.25, 0.75])
iqr = q3 - q1                     # the spread of the middle 50% of players

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print("Q1:", q1, "Q3:", q3, "IQR:", iqr)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
```

```python
# True if the wage is outside either fence, otherwise False
df['Wage_Outlier'] = (df['Wage'] < lower_fence) | (df['Wage'] > upper_fence)

print(df['Wage_Outlier'].value_counts())
print("Rows still in df:", len(df))
```
