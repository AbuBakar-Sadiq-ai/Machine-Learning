## Dataset

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
hours = np.array([1.5, 2.0, 2.5, 3.2, 4.0, 5.0, 5.5, 6.5, 7.2, 8.0])
scores = np.array([22.0, 28.0, 31.0, 40.0, 48.0, 56.0, 65.0, 72.0, 78.0, 88.0])
```

```python
def run_linear_regression(m, b, run_number):
    predictions = m * hours + b
    mse = np.mean((scores - predictions) ** 2)

    print("Run " + str(run_number) + " Results:")
    print("Slope (m): " + str(m))
    print("Intercept (b): " + str(b))
    print("Mean Squared Error: " + str(round(mse, 2)))
    print("-" * 30)

    plt.figure(figsize=(8, 5))
    plt.scatter(hours, scores, color="blue", label="Actual Student Scores")

    x_line = np.linspace(0, 10, 100)
    y_line = m * x_line + b
    plt.plot(x_line, y_line, color="black", label="Fit Line: y = " + str(m) + "x + " + str(b))

    for x_val, y_val, pred_val in zip(hours, scores, predictions):
        plt.plot([x_val, x_val], [y_val, pred_val], color="red", linestyle="--")

    plt.title("Run " + str(run_number) + " - Error (MSE): " + str(round(mse, 2)))
    plt.xlabel("Study Hours")
    plt.ylabel("Exam Score")
    plt.xlim(0, 10)
    plt.ylim(0, 105)
    plt.legend()
    plt.grid(True)
    plt.show()
```

## Run 1

Initial guess for the slope and intercept.

```python
run_linear_regression(m=10.3, b=5.5, run_number=1)
```

## Run 2

The slope is increased to make the line closer to the data points.

```python
run_linear_regression(m=9.0, b=5.0, run_number=2)
```

## Run 3

The slope is adjusted again to reduce the error.

```python
run_linear_regression(m=9.5, b=5.0, run_number=3)
```

## Run 4

The slope and intercept are adjusted to get a better fit.

```python
run_linear_regression(m=9.3, b=5.5, run_number=4)
```

## Run 5

Final adjustment for the line.

```python
# Calculate the best-fit line using least squares
m, b = np.polyfit(hours, scores, 1)

print("Best slope (m):", m)
print("Best intercept (b):", b)
print("Best fit line: y =", m, "x +", b)

run_linear_regression(m=m, b=b, run_number="Best Fit")
```
