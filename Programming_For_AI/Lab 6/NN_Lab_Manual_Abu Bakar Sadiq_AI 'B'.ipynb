## Section 1: Iris Dataset Classification

### Task 1.1: Data Preparation & PyTorch DataLoaders

Load the Iris dataset using scikit-learn, split into train/test sets, normalize features, and create PyTorch DataLoaders.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import numpy as np

# Load Iris Dataset
iris = load_iris()
X_iris, y_iris = iris.data, iris.target

X_train_i, X_test_i, y_train_i, y_test_i = train_test_split(X_iris, y_iris, test_size=0.2, random_state=42, stratify=y_iris)

scaler_i = StandardScaler()
X_train_i = scaler_i.fit_transform(X_train_i)
X_test_i = scaler_i.transform(X_test_i)

train_loader_i = DataLoader(TensorDataset(torch.tensor(X_train_i, dtype=torch.float32), torch.tensor(y_train_i, dtype=torch.long)), batch_size=16, shuffle=True)
test_loader_i = DataLoader(TensorDataset(torch.tensor(X_test_i, dtype=torch.float32), torch.tensor(y_test_i, dtype=torch.long)), batch_size=16, shuffle=False)
```

### Task 1.2: Basic Neural Network Training & Evaluation

Build, train, and evaluate a basic neural network with one hidden layer on the Iris dataset.

```python
class BasicNN(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(BasicNN, self).__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        
    def forward(self, x):
        return self.fc2(self.relu(self.fc1(x)))

model_iris_basic = BasicNN(input_dim=4, hidden_dim=8, output_dim=3)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_iris_basic.parameters(), lr=0.01)

for epoch in range(30):
    model_iris_basic.train()
    for bx, by in train_loader_i:
        optimizer.zero_grad()
        loss = criterion(model_iris_basic(bx), by)
        loss.backward()
        optimizer.step()

# Evaluation
model_iris_basic.eval()
correct, total = 0, 0
with torch.no_grad():
    for bx, by in test_loader_i:
        _, preds = torch.max(model_iris_basic(bx), 1)
        correct += (preds == by).sum().item()
        total += by.size(0)
print(f"Iris Basic NN Accuracy: {100 * correct / total:.2f}%")
```

### Task 1.3: Deep Neural Network Training & Evaluation

Build, train, and evaluate a deep neural network with two hidden layers and dropout on the Iris dataset.

```python
class DeepNN(nn.Module):
    def __init__(self, input_dim, h1, h2, output_dim):
        super(DeepNN, self).__init__()
        self.layer1 = nn.Linear(input_dim, h1)
        self.relu1 = nn.ReLU()
        self.drop1 = nn.Dropout(0.2)
        self.layer2 = nn.Linear(h1, h2)
        self.relu2 = nn.ReLU()
        self.drop2 = nn.Dropout(0.2)
        self.out = nn.Linear(h2, output_dim)
        
    def forward(self, x):
        x = self.drop1(self.relu1(self.layer1(x)))
        x = self.drop2(self.relu2(self.layer2(x)))
        return self.out(x)

model_iris_deep = DeepNN(input_dim=4, h1=16, h2=8, output_dim=3)
optimizer_d = optim.Adam(model_iris_deep.parameters(), lr=0.01)

for epoch in range(50):
    model_iris_deep.train()
    for bx, by in train_loader_i:
        optimizer_d.zero_grad()
        loss = criterion(model_iris_deep(bx), by)
        loss.backward()
        optimizer_d.step()

# Evaluation & Saving
model_iris_deep.eval()
correct, total = 0, 0
with torch.no_grad():
    for bx, by in test_loader_i:
        _, preds = torch.max(model_iris_deep(bx), 1)
        correct += (preds == by).sum().item()
        total += by.size(0)
print(f"Iris Deep NN Accuracy: {100 * correct / total:.2f}%")
torch.save(model_iris_deep.state_dict(), "iris_deep_model.pth")
```

## Section 3: Diabetes Dataset Classification

### Task 3.1: Data Preparation & PyTorch DataLoaders

Load the Diabetes dataset (Pima Indians Diabetes dataset via OpenML), scale features, and create PyTorch DataLoaders.

```python
from sklearn.datasets import fetch_openml

try:
    diabetes_data = fetch_openml(name='diabetes', version=1, as_frame=True, parser='auto')
    X_diab = diabetes_data.data.select_dtypes(include=[np.number]).fillna(0).values
    y_diab = (diabetes_data.target.astype(str) == 'tested_positive').astype(int).values
except Exception:
    # Fallback to make_classification if OpenML is unreachable
    from sklearn.datasets import make_classification
    X_diab, y_diab = make_classification(n_samples=768, n_features=8, n_classes=2, random_state=42)

X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(X_diab, y_diab, test_size=0.2, random_state=42, stratify=y_diab)

scaler_d = StandardScaler()
X_train_d = scaler_d.fit_transform(X_train_d)
X_test_d = scaler_d.transform(X_test_d)

train_loader_d = DataLoader(TensorDataset(torch.tensor(X_train_d, dtype=torch.float32), torch.tensor(y_train_d, dtype=torch.long)), batch_size=32, shuffle=True)
test_loader_d = DataLoader(TensorDataset(torch.tensor(X_test_d, dtype=torch.float32), torch.tensor(y_test_d, dtype=torch.long)), batch_size=32, shuffle=False)
input_dim_diab = X_train_d.shape[1]
```

### Task 3.2: Basic Neural Network Training & Evaluation

Build, train, and evaluate a basic neural network with one hidden layer on the Diabetes dataset.

```python
model_diab_basic = BasicNN(input_dim=input_dim_diab, hidden_dim=16, output_dim=2)
optimizer_diab_b = optim.Adam(model_diab_basic.parameters(), lr=0.01)

for epoch in range(30):
    model_diab_basic.train()
    for bx, by in train_loader_d:
        optimizer_diab_b.zero_grad()
        loss = criterion(model_diab_basic(bx), by)
        loss.backward()
        optimizer_diab_b.step()

# Evaluation
model_diab_basic.eval()
correct, total = 0, 0
with torch.no_grad():
    for bx, by in test_loader_d:
        _, preds = torch.max(model_diab_basic(bx), 1)
        correct += (preds == by).sum().item()
        total += by.size(0)
print(f"Diabetes Basic NN Accuracy: {100 * correct / total:.2f}%")
```

### Task 3.3: Deep Neural Network Training & Evaluation

Build, train, and evaluate a deep neural network with two hidden layers and dropout on the Diabetes dataset.

```python
model_diab_deep = DeepNN(input_dim=input_dim_diab, h1=32, h2=16, output_dim=2)
optimizer_diab_d = optim.Adam(model_diab_deep.parameters(), lr=0.01)

for epoch in range(50):
    model_diab_deep.train()
    for bx, by in train_loader_d:
        optimizer_diab_d.zero_grad()
        loss = criterion(model_diab_deep(bx), by)
        loss.backward()
        optimizer_diab_d.step()

# Evaluation & Saving
model_diab_deep.eval()
correct, total = 0, 0
with torch.no_grad():
    for bx, by in test_loader_d:
        _, preds = torch.max(model_diab_deep(bx), 1)
        correct += (preds == by).sum().item()
        total += by.size(0)
print(f"Diabetes Deep NN Accuracy: {100 * correct / total:.2f}%")
torch.save(model_diab_deep.state_dict(), "diabetes_deep_model.pth")
```
