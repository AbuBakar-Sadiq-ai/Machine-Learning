```python
b = data_x1_x2_target[:,2]
```

```python
b_arr = [1, 1, 1, 1]
```

```python
target = data_x1_x2_target[:,3]
```

```python
target = [ 1, -1, -1, -1]
```

```python
w1 = 0
w2 = 0
b = 0
```

```python
a = 1
```

```python
for j in range(20):
    print(f"epoch {j}: {w1},{w2},{b}")
    
    for i in range(4):
        
        t = target[i]
        
        input_y = w1 * x1[i] + w2 * x2[i] + b_arr[i]
        
        if input_y > 0:
            output_y = 1
        elif input_y < 0:
            output_y = -1
        else:
            output_y = 0
            
        w1tmp = w1 + a * x1[i] * t
        w2tmp = w2 + a * x2[i] * t
        btmp = b + a * t

        w1 = w1tmp
        w2 = w2tmp
        b = btmp
        print(f"{x1[i]=},{x2[i]=},{b_arr[i]=},{input_y=},{output_y=},{target=},{w1=},{w2=},{b=}")
```
