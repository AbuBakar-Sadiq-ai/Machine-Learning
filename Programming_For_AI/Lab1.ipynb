```python
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np

students = [
    {"Name": "Ali", "Age": "21", "Marks": "82", "Gender": "M"},
    {"Name": "Saad", "Age": "None", "Marks": "72", "Gender": "M"},
    {"Name": "Saba", "Age": "25", "Marks": "62", "Gender": "F"},
    {"Name": "Zain", "Age": "20", "Marks": "35", "Gender": "M"},
    {"Name": "Ahmad", "Age": "221", "Marks": "951", "Gender": "M"},
    {"Name": "Gulbadin", "Age": "31", "Marks": "56", "Gender": "M"},
    {"Name": "Shamshatto", "Age": "61", "Marks": "None", "Gender": "M"}
]
```

```python
df=pd.DataFrame(students)
print(df)
```

```python
print(df.dtypes)
```

```python
print("Handling data types :")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")  
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
print(df.dtypes)
```

```python
print("Missing values :")
df.isnull().sum()
```

```python
print("Handling missing values :")
df["Age"] = df["Age"].fillna(df["Age"].mean())
print("Missing values after handling :")
print(df.isnull().sum())
print(df)
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print("Missing values after handling :")
print(df.isnull().sum())
print(df)
```

```python
print("Outlier in Age:")
age_outliers = df[(df["Age"] > 100) | (df["Age"] < 15)]
print("Outliers :")
print(age_outliers)
```

```python
print("Outlier in Marks:")
marks_outliers = df[(df["Marks"] > 100) | (df["Marks"] < 0)]
print("Outliers :")
print(marks_outliers)
```

```python
# Replace outlier ages with the median (instead of dropping the row)
median_age = df.loc[~df.index.isin(age_outliers.index), "Age"].median()
df.loc[age_outliers.index, "Age"] = median_age

# Replace outlier marks with the median of valid marks
median_marks = df.loc[~df.index.isin(marks_outliers.index), "Marks"].median()
df.loc[marks_outliers.index, "Marks"] = median_marks

print("Cleaned outliers:")
print(df.loc[sorted(set(age_outliers.index) | set(marks_outliers.index))])
```

```python
# ---- Step 5: Clean categorical column (Gender) ----
df["Gender"] = df["Gender"].str.strip().str.upper()
```

```python
# ---- Step 6: Final cleaned dataset ----
df["Age"] = df["Age"].astype(int)  # Age should be a whole number
print("\nCleaned Data:")
print(df)
 
```

```python
df.to_csv("cleaned_data.csv", index=False)
print("\nCleaned data saved to 'cleaned_data.csv'")
```
