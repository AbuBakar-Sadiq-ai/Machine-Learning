```python
a=("pranam")
b=("GuruG")
```

```python
def cross_over(a, b):

    a_list = list(a)
    b_list = list(b)

    point = len(a_list) // 2
    
    offspring1 = a_list[:point] + b_list[point:]
    offspring2 = b_list[:point] + a_list[point:]
    
    offspring1_str = ''.join(offspring1)
    offspring2_str = ''.join(offspring2)
    
    return offspring1_str, offspring2_str
```

```python
print(cross_over(a, b))
```

```python
import random
def mutation(offspring, mutation_rate):
    mutated_offspring = []
    
    for individual in offspring:
        mutated_individual = list(individual)
        
        for i in range(len(mutated_individual)):
            if random.random() < mutation_rate:
                mutated_individual[i] = '1' if mutated_individual[i] == '0' else '0'
        
        mutated_offspring.append(''.join(mutated_individual))
    
    return mutated_offspring
```

```python
print(mutation(cross_over(a, b), 0.1))
```
