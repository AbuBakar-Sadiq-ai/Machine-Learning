```python
import random

TARGET = (34, 186, 56)        # (R, G, B) from the board
BITS_PER_CHANNEL = 8          # 0-255
CHROMOSOME_LENGTH = 3 * BITS_PER_CHANNEL   # 24 bits
POPULATION_SIZE = 100
GENERATIONS = 100
MUTATION_RATE = 0.02
CROSSOVER_RATE = 0.80

def create_chromosome():
    return [random.randint(0, 1) for _ in range(CHROMOSOME_LENGTH)]

def create_population():
    return [create_chromosome() for _ in range(POPULATION_SIZE)]

def decode_bits(bits):
    value = 0
    for bit in bits:
        value = value * 2 + bit
    return value

def decode(chromosome):
    r = decode_bits(chromosome[0:8])
    g = decode_bits(chromosome[8:16])
    b = decode_bits(chromosome[16:24])
    return r, g, b

def color_distance(color):
    r, g, b = color
    tr, tg, tb = TARGET
    return ((r - tr) ** 2 + (g - tg) ** 2 + (b - tb) ** 2) ** 0.5

def fitness(chromosome):
    cd = color_distance(decode(chromosome))
    return 1 / (1 + cd)

def roulette_selection(population, fitnesses):
    total = sum(fitnesses)
    r = random.random() * total
    cumulative = 0
    for chromosome, fit in zip(population, fitnesses):
        cumulative += fit
        if cumulative >= r:
            return chromosome[:]
    return population[-1][:]

def crossover(p1, p2):
    if random.random() > CROSSOVER_RATE:
        return p1[:], p2[:]
    point = random.randint(1, len(p1) - 1)
    return p1[:point] + p2[point:], p2[:point] + p1[point:]

def mutate(chromosome):
    child = chromosome[:]
    for i in range(len(child)):
        if random.random() < MUTATION_RATE:
            child[i] = 1 - child[i]
    return child

random.seed(42)
population = create_population()
best_overall, best_fit = None, -1

for generation in range(GENERATIONS):
    fitnesses = [fitness(c) for c in population]
    i = fitnesses.index(max(fitnesses))
    if fitnesses[i] > best_fit:
        best_overall, best_fit = population[i][:], fitnesses[i]
    print("Gen", generation, "Best RGB:", decode(population[i]), "Fitness:", round(fitnesses[i], 4))
    if best_fit == 1.0:
        break

    new_population = [best_overall[:]]          # elitism: keep the best animal
    while len(new_population) < POPULATION_SIZE:
        p1 = roulette_selection(population, fitnesses)
        p2 = roulette_selection(population, fitnesses)
        c1, c2 = crossover(p1, p2)
        new_population.append(mutate(c1))
        if len(new_population) < POPULATION_SIZE:
            new_population.append(mutate(c2))
    population = new_population

print("Target RGB:", TARGET)
print("Best RGB  :", decode(best_overall))
print("Fitness   :", round(best_fit, 4))
```
